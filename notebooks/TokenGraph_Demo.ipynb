## Create a small project

```python
from pathlib import Path

root = Path("/content/support_project")
(root / "prompts").mkdir(parents=True, exist_ok=True)
(root / "prompts" / "system.txt").write_text(
    "You are a support assistant. Return only JSON with intent and confidence.",
    encoding="utf-8",
)
(root / "app.py").write_text(
    '''from pathlib import Path

SYSTEM_PROMPT = Path("prompts/system.txt").read_text(encoding="utf-8")

PROMPT = f"""{SYSTEM_PROMPT}\nHistory: {conversation_history}\nMessage: {message}"""

def classify(client, conversation_history, message):
    return client.chat.completions.create(prompt=PROMPT, messages=conversation_history)
''',
    encoding="utf-8",
)
print(root)
```

## Build the graph

```python
!tokenoptipy build /content/support_project --output /content/tokenoptipy-out
```

## Inspect statistics and hotspots

```python
!tokenoptipy stats --graph /content/tokenoptipy-out/graph.json
!tokenoptipy hotspots --graph /content/tokenoptipy-out/graph.json
```

## Search and explain nodes

```python
!tokenoptipy query "history" --graph /content/tokenoptipy-out/graph.json
!tokenoptipy explain conversation_history --graph /content/tokenoptipy-out/graph.json
```
