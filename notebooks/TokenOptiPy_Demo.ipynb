```python
from tokenoptipy import ValidationPolicy, optimize_prompt

prompt = """
Please note that it is important to classify the message.
Classify the message.
Classify the message.
Return JSON only.
Customer: {{customer_message}}
"""

report = optimize_prompt(
    prompt,
    backend="auto",
    policy=ValidationPolicy(required_terms=("JSON",)),
)

print(report.selected_prompt)
print(report.to_dict())
```
